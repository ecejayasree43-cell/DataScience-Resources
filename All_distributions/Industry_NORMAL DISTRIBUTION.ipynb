```python
"""
NORMAL DISTRIBUTION --->There is a mean and standard deviation, and values tend to cluster around the mean.
That's NORMAL

For Normal Distribution, easiest method:

from scipy.stats import norm
Formula:
Z = (X - μ) / σ
"""
```

```python
'''
Problem 1 — Smartphone Battery Life 
Battery life is normally distributed with:
•	Mean = 10 hours
•	SD = 1.5 hours
What is the probability that a randomly selected phone lasts more than 12 hours?
'''
from scipy.stats import norm

mean = 10
sd = 1.5
x = 12

z = (x - mean) / sd

probability = 1 - norm.cdf(z)

print("Z =", z)
print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 2 — Employee Commute Time 
Employee commute time follows a normal distribution.
•	Mean = 35 minutes
•	SD = 8 minutes
What is the probability that an employee's commute takes between 30 and 45 minutes?
'''
from scipy.stats import norm

mean = 35
sd = 8

x1 = 30
x2 = 45

z1 = (x1 - mean) / sd
z2 = (x2 - mean) / sd

probability = norm.cdf(z2) - norm.cdf(z1)

print("Z1 =", z1)
print("Z2 =", z2)
print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```
