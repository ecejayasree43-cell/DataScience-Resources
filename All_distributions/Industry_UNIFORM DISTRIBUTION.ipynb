```python
'''
Formula:
P(a < X < b) = (b - a) / (High - Low)
For continuous Uniform Distribution.

'''
```

```python
'''
Problem 1 — Random Number Generator 
A random number generator produces a number uniformly between 1 and 50.
What is the probability that the number is between 20 and 35?
'''

#Random number between 1 and 50 → probability between 20 and 35
low = 1
high = 50

a = 20
b = 35

probability = (b - a) / (high - low)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 2 — Bus Waiting Time 
A bus arrives randomly between 0 and 30 minutes after a passenger arrives.
What is the probability that the passenger waits less than 10 minutes?
'''

#Bus arrives randomly between 0 and 30 min → probability < 10 min
low = 0
high = 30

x = 10

probability = (x - low) / (high - low)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 3 — Random Game Score 
A game generates a score uniformly between 0 and 100.
What is the probability that the score is above 75?
'''

#Score uniformly distributed from 0–100 → probability >75
low = 0
high = 100

x = 75

probability = (high - x) / (high - low)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 4 — Random Appointment Time 
A doctor may call a patient randomly between 2:00 PM and 3:00 PM.
What is the probability that the patient is called before 2:15 PM?
'''

'''
Doctor calls randomly between 2 PM and 3 PM → probability before 2:15 PM
Convert everything into minutes:
2 PM = 0 minutes
2:15 PM = 15 minutes
3 PM = 60 minutes
'''
low = 0
high = 60

x = 15

probability = (x - low) / (high - low)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 5 — Sensor Measurement 
A sensor produces a reading uniformly between 100 and 200 units.
What is the probability that the reading falls between 140 and 170?
'''
#Sensor reading uniformly distributed from 100–200 → probability between 140–170
low = 100
high = 200

a = 140
b = 170

probability = (b - a) / (high - low)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```
