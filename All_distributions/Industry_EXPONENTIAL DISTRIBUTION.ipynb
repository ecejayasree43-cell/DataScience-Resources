```python
'''
EXPONENTIAL DISTRIBUTION --->how long to wait for event occurs 
Remember:
λ = 1 / Mean Waiting Time
For within x:
P(X ≤ x) = 1 - e^(-λx)
For after x:
P(X > x) = e^(-λx)
'''
```

```python
'''
Problem 1 — Taxi Arrival 
A taxi arrives at a location once every 15 minutes on average.
What is the probability that the next taxi arrives within 5 minutes?
'''
import math

mean = 15
x = 5

lam = 1 / mean

probability = 1 - math.exp(-lam * x)

print("Lambda =", lam)
print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```
