```python
'''
Problem 1 — IT Support Tickets 
An IT help desk receives an average of 8 support tickets per hour.
What is the probability that exactly 5 tickets arrive in the next hour?
'''
import math

lam = 8
k = 5

probability = math.exp(-lam) * lam**k / math.factorial(k)

print(probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 2 — Restaurant Customers 
A restaurant receives an average of 12 customers every 30 minutes.
What is the probability that exactly 15 customers arrive during the next 30 minutes?
'''
import math

lam = 12
k = 15

probability = math.exp(-lam) * lam**k / math.factorial(k)

print(probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 3 — Server Errors 
A server experiences an average of 3 errors per hour.
What is the probability of observing more than 4 errors in the next hour?
'''
import math
lam = 3
probability = 0

for k in range(5):
    probability += math.exp(-lam) * lam**k / math.factorial(k)

result = 1 - probability

print(result)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 4 — Emergency Calls 
An emergency center receives an average of 6 calls per hour.
What is the probability that zero calls are received during the next 20 minutes?
'''
import math
lam = 2

probability = math.exp(-lam)

print(probability)
print("Percentage =", probability * 100, "%")
```
