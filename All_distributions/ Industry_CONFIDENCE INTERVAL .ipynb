```python
'''
CONFIDENCE INTERVAL ---> To find What range probably contains the true average 

Formula:
CI=xˉ±Zσ
Useful Z values:
Confidence	Z
90%	1.645
95%	1.96
99%	2.576

For these problems:
CI = Mean ± Z × (SD / √n)
Import:   import math


note:
For these problems, assume the given standard deviation is the population
standard deviation and use the appropriate Z critical value.
'''
```

```python
'''
Problem 1 — Smartphone Battery Life 
A company tests 64 smartphones.
The average battery life is 18 hours, with a standard deviation of 2 hours.
Find the 95% confidence interval for the true average battery life.
'''
import math

mean = 18
sd = 2
n = 64
z = 1.96

se = sd / math.sqrt(n)
margin = z * se

lower = mean - margin
upper = mean + margin

print("Standard Error =", se)
print("Margin of Error =", margin)
print("95% Confidence Interval =", lower, "to", upper)

```

```python
'''
Problem 2 — Delivery Time 
A delivery company studies 100 deliveries.
Mean delivery time = 42 minutes
Standard deviation = 8 minutes
Find the 99% confidence interval for the true average delivery time.
'''
import math

mean = 42
sd = 8
n = 100
z = 2.576

se = sd / math.sqrt(n)
margin = z * se

lower = mean - margin
upper = mean + margin

print("99% Confidence Interval =", lower, "to", upper)
```
