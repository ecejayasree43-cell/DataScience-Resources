```python
'''
BINOMIAL DISTRIBUTIONS --> To find successrate out of fixed rate (based on trails and events)
from scipy.stats import binom
The important functions are:
binom.pmf() → Exactly
binom.cdf() → At most (≤)
binom.sf()  → Greater than (>)
'''
```

```python
'''
POINTS TO REMEMEBER:
"EXACTLY"       → binom.pmf()

"AT MOST 5"     → binom.cdf(5)

"LESS THAN 5"   → binom.cdf(4)

"AT LEAST 6"    → binom.sf(5)

"MORE THAN 5"   → binom.sf(5)
One small thing to remember: sf(x) means P(X > x), so for "at least 6", use sf(5), not sf(6). This is a very common beginner mistake. 
'''
```

```python
'''
Problem 1 — Manufacturing Quality Control 
A factory produces electronic chips. Historical data shows that 98% of the chips pass quality inspection.
If 12 chips are randomly selected, what is the probability that exactly 11 chips pass the inspection?
'''


from math import comb
print("============METHOD-1===================")
n = 12
k = 11
p = 0.98

probability = comb(n, k) * p**k * (1-p)**(n-k)

print(probability)
print("Percentage =", probability * 100, "%")

```

```python
from scipy.stats import binom
print("==============METHOD-2============================")
n = 12
p = 0.98
k = 11

probability = binom.pmf(k, n, p)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
'''
Problem 2: 75% of students complete an online course.
If 8 students enroll, what is the probability that at least 6 complete?


At least 6 means:  P(X≥6)
We can calculate:
P(X≥6)=1−P(X≤5)
'''
from scipy.stats import binom

n = 8
p = 0.75

probability = binom.sf(5, n, p)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")

```

```python
'''
Problem3: A medical test correctly identifies a patient's condition 90% of the time.
For 7 patients, what is the probability that it correctly identifies exactly 6?
'''
from scipy.stats import binom

n = 7
p = 0.90
k = 6

probability = binom.pmf(k, n, p)

print("Probability =", probability)

print("Percentage =", probability * 100, "%")

```

```python
'''
Problem 4: An online advertisement has a 30% click probability.
If it is shown to 10 people, what is the probability that exactly 4 people click?
'''
from scipy.stats import binom

n = 10
p = 0.30
k = 4

probability = binom.pmf(k, n, p)

print("Probability =", probability)
print("Percentage =", probability * 100, "%")

```
